```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import warnings

warnings.filterwarnings('ignore', category=UserWarning, module='openpyxl')

# Data inladen
bestand = "/Users/hbraakmann/Desktop/QGIS_DATA/Monitoring_Broedvogels/trends_watervogels_2022-2023.xlsx"


df = pd.read_excel(bestand, sheet_name="seizoensgetallen Nederland", skiprows=2)
df.columns = df.columns.astype(str)

soorten_watervogels = ["Grutto", "Tureluur", "Slobeend", "Kluut"]

df_landelijk = df.loc[df["Soort"].isin(soorten_watervogels)]
df_landelijk = df_landelijk[["Soort"] + [str(y) for y in range(1975, 2023)] + ["verandering vanaf 2011"]].reset_index(drop=True)
df_landelijk
```

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))
axes = axes.flatten()

for index, soort in df_landelijk.iterrows():
    ax = axes[index]
    
    soortnaam = soort["Soort"]
    jaren = soort.index[1:-1]
    waarden = soort.iloc[1: -1].values
    
    base_value = waarden[-12]  # 12 jaar
    trend = soort.iloc[-1]
    trend_line = [base_value * (1 + trend/100) ** i for i in range(12)]
    
    ax.plot(jaren, waarden, color="skyblue", marker="o", markersize=5, mfc="yellow")
    ax.plot(jaren[-12:], trend_line, color="red", linestyle="--", alpha=0.7, label=f"Trend 2011-2022: {trend:+.1f}% p.j.")
    ax.tick_params(axis="x", rotation=90, labelsize=8, colors="dimgrey")
    ax.tick_params(axis="y", labelsize=8, colors="dimgrey")
    ax.set_xticks(jaren[::5])
    ax.set_xlim(jaren.min(), jaren.max())
    ax.grid(axis="both", alpha=0.3) 
    ax.set_title(soortnaam, color="dimgrey")
    ax.spines[["top","right"]].set_visible(False)
    ax.spines[["left","bottom"]].set_color("darkgrey")
    ax.legend(loc="best", fontsize=8, framealpha=0.9)

    # Y-as label
    if index % 2 == 0:  # Linker kolom
        ax.set_ylabel("Aantal", color="dimgrey")
    if index >= 2:  # Onderste rij
        ax.set_xlabel("Jaar", color="dimgrey")

fig.suptitle("Landelijk ontwikkeling watervogels 1975–2022", color="darkgrey", fontsize=16, y=1)
plt.tight_layout()
plt.show()
```
